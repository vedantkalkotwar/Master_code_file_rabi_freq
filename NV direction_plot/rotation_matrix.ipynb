```python
import sympy as sp


def rotation_to_z(v):
    """Exact (symbolic) rotation matrix R such that R @ v = |v| * [0, 0, 1]."""
    v = sp.Matrix(v)
    norm = sp.sqrt(sp.simplify(v.dot(v)))
    if sp.simplify(norm) == 0:
        raise ValueError("Zero vector has no direction.")
    v_hat = (v / norm).applyfunc(sp.simplify)
    z = sp.Matrix([0, 0, 1])

    k = v_hat.cross(z)                 # rotation axis, |k| = sin(theta)
    c = sp.simplify(v_hat.dot(z))      # cos(theta)

    # Edge cases: v parallel / antiparallel to z
    if all(sp.simplify(ki) == 0 for ki in k):
        if sp.simplify(c) > 0:
            return sp.eye(3)
        return sp.diag(1, -1, -1)      # 180 deg rotation about x-axis

    K = sp.Matrix([[0, -k[2], k[1]],
                   [k[2], 0, -k[0]],
                   [-k[1], k[0], 0]])

    # Rodrigues: R = I + K + K^2 / (1 + c)   [since (1-c)/s^2 = 1/(1+c)]
    R = sp.eye(3) + K + K * K / (1 + c)
    return R.applyfunc(lambda e: sp.radsimp(sp.simplify(e)))


if __name__ == "__main__":
    v = sp.Matrix([-1*sp.sqrt(sp.Rational(2,3)),sp.sqrt(sp.Rational(2,9)),sp.Rational(-1,3)])

    R = rotation_to_z(v)

    print("Rotation matrix R:")
    sp.pprint(R)
    print("\nR @ v =");            sp.pprint((R * v).applyfunc(sp.simplify))
    print("\nR @ R.T =");          sp.pprint((R * R.T).applyfunc(sp.simplify))
    print("\ndet(R) =", sp.simplify(R.det()))
    print("\nLaTeX:\n", sp.latex(R))
```
